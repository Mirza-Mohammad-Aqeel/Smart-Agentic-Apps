```python
import sys
import os
from pathlib import Path
import streamlit as st
import shutil

root_dir = Path(__file__).resolve().parent.parent

if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

STORAGE_DIR = Path(root_dir / "Photos Gallery")

if STORAGE_DIR.exists():
    shutil.rmtree(STORAGE_DIR)

STORAGE_DIR.mkdir(exist_ok=True)

st.set_page_config(page_title="Smart Photo Gallery", layout="centered")

st.title("Smart Photo Gallery")

tab_online, tab_offline = st.tabs(["Upload Photos", "Manage Gallery Locker"])

with tab_online:
    st.subheader("Upload New Photos")
    
    uploaded_files = st.file_uploader(
        "Choose Images to Upload", 
        type=["png", "jpg", "jpeg"], 
        accept_multiple_files=True,
        key="photo_uploader"
    )
    
    if st.button("Save Uploaded Photos to Locker", type="primary"):
        if not uploaded_files:
            st.warning("Please Select a Valid Image")
        else:
            saved_count = 0
            for uploaded_file in uploaded_files:
                target_path = STORAGE_DIR / uploaded_file.name
                
                with open(target_path, "wb") as f:
                    f.write(uploaded_file.getbuffer())
                saved_count += 1
                
            st.success(f"Successfully Saved {saved_count} Images to your Locker!")

with tab_offline:
    st.subheader("Your Photo Gallery Locker")
    
    current_photos = [f for f in os.listdir(STORAGE_DIR) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    
    if not current_photos:
        st.info("Your Gallery Locker is Currently Empty, Upload Photos!")
    else:
        st.write(f"Total Photos Stored = {len(current_photos)} Photos")
        st.markdown("---")
        
        for photo_name in current_photos:
            photo_path = STORAGE_DIR / photo_name
            
            with st.container():
                col1, col2 = st.columns([3, 1])
                
                with col1:
                    st.image(str(photo_path), caption=photo_name, use_container_width=True)
                
                with col2:
                    st.write("") 
                    st.write("")
                    if st.button(f"🗑️ Delete", key=f"del_{photo_name}", type="secondary"):
                        try:
                            os.remove(photo_path)
                            st.success(f"Deleted '{photo_name}'!")
                            st.rerun() 
                        except Exception as e:
                            st.error(f"Failed to Delete File")
            st.markdown("---")
```
